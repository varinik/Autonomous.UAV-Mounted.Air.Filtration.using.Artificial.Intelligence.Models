```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Sample 3D data
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z_main = np.sin(np.sqrt(X**2 + Y**2))
Z1 = np.cos(X)
Z2 = np.sin(Y)
Z3 = np.exp(-0.1 * (X**2 + Y**2))

# Setup figure and grid layout
fig = plt.figure(figsize=(12, 8))
gs = gridspec.GridSpec(2, 3, height_ratios=[2, 1.25])

# Main big 3D plot
ax_main = fig.add_subplot(gs[0, :], projection='3d')
ax_main.plot_surface(X, Y, Z_main, cmap='viridis')
ax_main.set_title("Main 3D Plot")

# Smaller 3D subplots
ax1 = fig.add_subplot(gs[1, 0], projection='3d')
ax1.plot_surface(X, Y, Z1, cmap='plasma')
ax1.set_title("Subplot 1")

ax2 = fig.add_subplot(gs[1, 1], projection='3d')
ax2.plot_surface(X, Y, Z2, cmap='inferno')
ax2.set_title("Subplot 2")

ax3 = fig.add_subplot(gs[1, 2], projection='3d')
ax3.plot_surface(X, Y, Z3, cmap='cividis')
ax3.set_title("Subplot 3")

plt.tight_layout(pad=20)

plt.show()
```
